```python
import sqlite3 as sql3
```

```python
# Gerando banco de dados com sql3
conexao = sql3.connect('titulo.db')
```

```python
# Gerando cursor para execução de querys em sql
cursor = conexao.cursor()
```

```python
# Criando tabela com query no cursor
cursor.execute(
    '''
        CREATE TABLE filmes(
        id INTEGER NOT NULL PRIMARY KEY AUTOINCREMENT,
        nome TEXT NOT NULL,
        ano INTEGER NOT NULL,
        nota REAL NOT NULL
        )
    '''
)
```

```python
# Fechendo a conexão para evitar inconsistências
conexao.close()
print("Tabela criada!")

'''
Por padrão, sempre é necessário encerrar a conexão de dados, você pode excerra a conexão do cursor, se ainda
for usar o cursor para execulta outra query. Se você encerrar a conexão, não podera declarar e usar cursores.
'''
```

```python
# Dropando base de dados(tabela) com nova estrutura
with sql3.connect('titulo.db') as conexao:
    cursor = conexao.cursor()
    cursor.execute(
        '''
            DROP TABLE filmes
        '''
    )
conexao.commit()
conexao.close()
```
