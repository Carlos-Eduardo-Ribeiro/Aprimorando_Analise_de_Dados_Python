```python
import sqlite3 as sql3
import pandas as pd
```

```python
# Gerando conexão e criando cursor
conexao = sql3.connect('titulo.db')
cursor = conexao.cursor()
```

### Realiando leitura com sql3

```python
# Extraindo dados da base de dados e atribuindo o resultado a uma variável
dados =  cursor.execute("SELECT * FROM filmes")

# Fechando conexão
conexao.close()

# Somente é necessário usar a conexao.commit() em situações de alterações no banco ou seja, quase tudo, menos na leitura.
```

```python
# Convertendo o obj para uma lista de tuplas com fetchall
print(dados.fetchall())
```

### Realizando leitura com pandas(BEM MELHOR)

```python
# Gerando query a atribuindo a uma variável
query = "SELECT * FROM filmes"
```

```python
# Usando pandas para ler o banco de dados em gerando um DF
df = pd.read_sql(query, conexao)
```
