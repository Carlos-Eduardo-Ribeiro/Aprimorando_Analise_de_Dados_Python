```python
import sqlite3 as sql3
```

```python
# Gerando conexão e criando cursor
conexao = sql3.connect('titulo.db')
cursor =  conexao.cursor()
```

```python
# Inserindo dados na tabela filmes com o cursor
cursor.execute(
    '''
        INSERT INTO filmes(nome, ano, nota)
        VALUES
            ("Era do gelo", 2015, 9.75),
            ("A Torre negra", 2020, 7.5),
            ("Tubarão assassino", 1999, 9.30)
    '''
)
```

```python
# Subindo dados para o db e encerrando a conexão
conexao.commit()
conexao.close()
print("Dados inseridos na base de dados!")
```
