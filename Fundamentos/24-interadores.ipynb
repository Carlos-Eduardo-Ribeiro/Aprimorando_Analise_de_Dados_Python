```python
class Contador:

    def _init_(self, Limite):
        self.limite = Limite
        self.atual = 1
    
    def _iter_(self):
        return self
    
    def _next_(self):
        if self.atual > self.limite:
            raise StopIteration
        else:
            valor  = self.atual
            self.atual +=1
            return valor
```
